```python
import pandas as pd
import csv
```

```python
#read frequency csv into a dataframe
freq_path = ("C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\PRG1978\\constructions\\constructicon_workflow"
             "\\001_morphosyntactic_extraction\\results\\frequency_lists\\elative_construction_lemma_freqs.csv")
freq_df = pd.read_csv(freq_path)
freq_df
```

```python
#find the frequency of each lemma in member_1
lemma_freq = {}

for index, row in freq_df.iterrows():
    lemma = row['member_1']
    freq = row['frequency']
    if lemma not in lemma_freq.keys():
        lemma_freq[lemma] = freq
    else:
        lemma_freq[lemma] += freq
```

```python
#save lemmas with a frequency with 5 or above into a list
lemmafreq_small = []

for lemma, freq in lemma_freq.items():
    if freq > 4:
        lemmafreq_small.append(lemma)
```

```python
#save the list into a csv file where each lemma is on its own row
smallfreq_path = ("C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\PRG1978\\constructions\\constructicon_workflow"
             "\\001_morphosyntactic_extraction\\results\\wordlist_freq5.csv")
with open(smallfreq_path, 'w', newline='', encoding="utf-8") as f:
    writer = csv.writer(f)
    for lemma in lemmafreq_small:
        writer.writerow([lemma])
```
