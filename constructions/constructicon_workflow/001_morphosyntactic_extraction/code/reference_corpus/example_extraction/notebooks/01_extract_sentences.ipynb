## Extracting sentences from Estonian Reference Corpus 

#### I Accessing corpus, choosing collection and layers

```python
from estnltk import Text
from estnltk.storage.postgres import PostgresStorage

storage = PostgresStorage(pgpass_file='pgpass.txt', 
                          schema='estonian_text_corpora', 
                          role='estonian_text_corpora_read')

storage
```

```python
collection = storage['koondkorpus_sentences']
```

Stanza syntax layer is selected, as it isn't among the default layers.

```python
collection.selected_layers = ['v172_stanza_syntax']
```

#### II Extracting sentences that contain candidates for elative construction examples

```python
import sys

sys.path.append("..")

import time


# Collecting sentences and their indexes
extracted_examples = []
collection_indexes = []

start_time = time.time()

for sent_idx, sent_obj in enumerate(collection):

    words = sent_obj.v172_stanza_syntax
    
    for stanza_word in words:
        
        # Early stopping if word is not a nominal modifier
        if stanza_word.deprel != "nmod":
            continue

        morph = stanza_word.morph_analysis
        pos = morph.partofspeech
        form = morph.form

        # Early stopping if nominal modifier is not a substantive or isn't in singular elative case
        if "S" not in pos or "sg el" not in form:
            continue
       
        parent = stanza_word.parent_span

        # Early  stopping if nominal modifier doesn't have a syntactic parent
        if parent is None:
            continue
        
        parent_pos = parent.morph_analysis.partofspeech

        # Early stopping if parent is not a substantive or proper noun
        if not ("S" in parent_pos or "H" in parent_pos):
            continue
        
        # saving sentence Text-object and its index
        extracted_examples.append(sent_obj)
        collection_indexes.append(sent_idx)
        break

print("--- %s seconds ---" % (time.time() - start_time))

storage.close()

from estnltk_core.converters.json_exporter import text_to_json

# Saving extracted sentences in JSON-format
output_dir = "../data/koondkorpus_sentences/conx_example_sentences_20042026/"

for idx, example in enumerate(extracted_examples):
    example.meta["sentence_id"] = collection_indexes[idx]
    text_to_json(example, f"{output_dir}{collection_indexes[idx]}.json")
```
