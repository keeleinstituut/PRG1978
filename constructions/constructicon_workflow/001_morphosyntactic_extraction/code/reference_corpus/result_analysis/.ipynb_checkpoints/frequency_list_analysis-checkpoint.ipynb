# Frequncy of examples and collexemes

```python
import pandas as pd
```

```python
#read frequency csv into a dataframe
freq_path = ("C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\PRG1978\\constructions\\constructicon_workflow"
             "\\001_morphosyntactic_extraction\\results\\frequency_lists\\elative_construction_lemma_freqs.csv")
freq_df = pd.read_csv(freq_path)
freq_df
```

```python
#find the frequency of each lemma in member_1
lemma_freq = {}

for index, row in freq_df.iterrows():
    lemma = row['member_1']
    freq = row['frequency']
    if lemma not in lemma_freq.keys():
        lemma_freq[lemma] = freq
    else:
        lemma_freq[lemma] += freq
```

```python
print("Unique lemmas in component one (noun in elative): ", len(lemma_freq))
```

```python
lemmafreq_small = {}
freq_sum = 0 #sum on all the lemma frequencies
freq_sum_small = 0 #sum of all lemma frequencies above 5
hapax = 0
for lemma, freq in lemma_freq.items():
    freq_sum += freq
    if freq > 4:
        freq_sum_small += freq
        lemmafreq_small[lemma] = freq
    if freq == 1:
        hapax += 1
```

```python
print("Unique lemmas in component one (noun in elative) with frequency 5 and above: ", len(lemmafreq_small))
print("Frequency of the morphosyntactic pattern in the corpus: ", freq_sum)
print("How many appearances of the morphosyntactic pattern are excluded when we only look at lemmas with freq 5 or above: ",
      freq_sum-freq_sum_small, " or ", round((freq_sum-freq_sum_small)/freq_sum*100,2), "%")
print("Amount of hapax legomenon in component 1: ", hapax)
```

```python
df_lemmafreq = pd.DataFrame.from_dict(lemma_freq, orient="index", columns=["frequency"])
df_lemmafreq = df_lemmafreq.sort_values(by=["frequency"], ascending=False)
df_lemmafreq
```

```python
top2000_freq = df_lemmafreq.head(2000).sum().item()
print("How many appearances of the morphosyntactic pattern are excluded when we only look at the top 2000 lemmas by frequency: ",
      freq_sum-top2000_freq, " or ", round((freq_sum-top2000_freq)/freq_sum*100,2), "%")
```

```python
df_lemmafreq.to_csv("member1_lemma_freq.csv", index_label = "lemma")
```
