## Extracting elative construction instances from Estonian Reference Corpus

 - Extracting elative construction pattern matches from pre-filtered Estonian Reference Corpus sentences.
 - Semantically filtering construction members in elative case and allocating found pattern matches to categories of material and occupation/nationality/role.
 - Finding 10 most common and most attracted construction member lemmas in elative case for both categories.
 - Finding 5 most common elative construction phrases per each lemma for both categories.
 - Finding 3 example sentences per each phrase for both categories.
 - Saving results in CSV-format.

Later, these results will be exported to EKILEX database.

### II Extracting elative construction pattern matches

```python
# setting up pattern of elative construction
el_conx_pattern = {
        "nodes": {
            "A": {
                "deprels": ["nmod"], # A has to be a substantive and a nominal modifier in singular elative case
                "partofspeeches": ["S"],
                "forms": ["sg el"],
                "lemmas": None,
                "lemma_regex": None,
                "instances": None
                },
            "B": {
                "deprels": None,
                "partofspeeches": ["S"], # B has to be a substantive
                "forms": None,
                "lemmas": None,
                "lemma_regex": None,
                "instances": None
                }
        },
        "node_restrictions": {
            "A": None,
            "B": {"deprels": ["nmod"], # B can't be a nominal modifier, in terminative case, nor can it's lemma end with "mine"-clitic
                  "partofspeeches": None,
                  "forms": ["sg ter", "pl ter"],
                  "lemmas": None,
                  "lemma_regex": [r".*mine$"],
                  "instances": None,
                  "parent": None,
                  "child": None
                }
        },
        "edges": [
            ("B", "A")  # B is head of A
        ],
        "order": [
            ("A", "BEFORE", "B") # A has to appear before B in text
        ]
    }
```

```python
# finding elative construction instances from Estonian Reference Corpus sentences based on pattern
match_sentences = []
matches = []

for idx, sentence in enumerate(data):

    sent_matches = get_pattern(stanza_words = sentence.v172_stanza_syntax, 
                                pattern = el_conx_pattern, 
                                anchor_name = "B")
    if len(sent_matches) > 0:
        match_sentences.append(sentence.v172_stanza_syntax)
        matches.append(sent_matches)
```

### III Preparing lemmas and form codes of members in elative case from extracted pattern matches for semantic filtering and computing statistics

```python
# list of syntactic heads of construction examples
conx_heads_list = [[get_conx_head(sent, match) for idx2, match in enumerate(matches[idx1])] for idx1, sent in enumerate(match_sentences)]
print(conx_heads_list[0])

# list of specified members of construction examples
conx_member_list = [[get_conx_member(match, "A") for idx2, match in enumerate(matches[idx1])] for idx1, sent in enumerate(match_sentences)]
print(conx_member_list[0])

# list of specified members of construction examples as lemmas
conx_member_lemma_list = [[member.morph_analysis.lemma[0] for member in members] for idx1, members in enumerate(conx_member_list)]
print(conx_member_lemma_list[0])

# flattened version of member lemma list
lemma_list = [member for members in conx_member_lemma_list for member in members]
print(lemma_list[0])

# list of specified members of construction examples as form codes
conx_member_form_codes_list = [[get_form_code(member) for member in members] for idx1, members in enumerate(conx_member_list)]
print(conx_member_form_codes_list[0])

# flattened version of member form codes list
form_codes = [member for members in conx_member_form_codes_list for member in members]
print(form_codes[0])
```

### V Elative material

```python
# list of construction members in elative case as EKILEX form codes
el_material_conx_member_form_codes_list = [[get_form_code(member) for member in members if member.morph_analysis.lemma[0] in elative_material] for idx1, members in enumerate(conx_member_list)]
el_material_conx_member_form_codes_list_cleaned = [el for el in el_material_conx_member_form_codes_list if len(el) > 0]
print(el_material_conx_member_form_codes_list_cleaned[0])

# flattened version of member form codes list
el_material_form_codes = [member for members in el_material_conx_member_form_codes_list_cleaned for member in members]
print(el_material_form_codes[0])
```

```python
# calculating statistics, getting n most frequent and n most attracted form code lemmas
# n most common ja n most attracted frequency- ja LL-scores
el_material_most_common, el_material_most_attracted = get_statistics(all_member_forms=el_material_form_codes, 
                                                      conx_freq=len(el_material_form_codes),
                                                      n_most_common=10,
                                                      n_most_attracted=10)
```

```python
print(el_material_most_common)
print(el_material_most_attracted)
```

```python
# extracting examples for n most frequent and n most attracted lemmas
material_top10 = el_material_most_common | el_material_most_attracted

material_n_most_common_phrases = dict()
material_n_most_common_phrases_stanza = dict()

for lemma in material_top10.keys():
    phrase_examples_str, phrase_exaples_stanza = extract_n_most_common_phrases(lemma, 5, "A", match_sentences, matches)
    material_n_most_common_phrases[lemma] = phrase_examples_str
    material_n_most_common_phrases_stanza[lemma] = phrase_exaples_stanza

material_phrase_example_sentences = dict()

for example_list in material_n_most_common_phrases.values():
    for example in example_list:
        phrase_sentences = extract_example_sentences(example, 3, match_sentences, matches)
        material_phrase_example_sentences[example] = phrase_sentences
```

```python
material_n_most_common_phrases
```

Saving results in CSV-format:

```python
# elative material construction
top10_material = el_material_most_attracted | el_material_most_common

material_data = []

for word, stats in top10_material.items():
    for idx1, phrase in enumerate(material_n_most_common_phrases[word]):
        for idx2, sentence in enumerate(material_phrase_example_sentences[phrase]):
            data_dct = {
                        "lemma": word, 
                        "frequency": stats[0], 
                        "salience": stats[1], 
                        "phrase": phrase, 
                        "sentence": " ".join([word.text for word in sentence])
                        }
            material_data.append(data_dct)
```

```python
material_data[:10]
```

```python
with open('../../../../results/reference_corpus_data/elative_construction/elative_material_construction.csv', 'w', encoding='utf-8', newline='') as csvfile:
    fieldnames = ['lemma', 'frequency', 'salience', 'phrase', 'sentence']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(material_data)
```
