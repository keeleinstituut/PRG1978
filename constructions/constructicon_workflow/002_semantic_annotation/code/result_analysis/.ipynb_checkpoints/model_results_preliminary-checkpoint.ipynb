```python
import pandas as pd
import numpy as np
from sklearn.metrics import f1_score
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
```

```python
def experiment_big(gold_df, model_df, model1, model2, column_gold):
    model_results = []

    for value in gold_df[column_gold]: #iterate over gold standard rows and take specified column's value from each row
        annotation_attribute = model_df[model1].loc[value]['elukutse/rahvus/roll'] #take LLM's annotation for this value from elative_attribute experiment
        #average is taken for when the same word/phrase/sentence occurs multiple times
        if np.average(annotation_attribute) == 1:  #if LLM annotated the value as being an occupation, nationality or role
            model_results.append('elative_attribute') #tag the whole row as being an instance of the elative_attribute construction
        else: #if LLM annotated value as NOT being an occupation, nationality or role
            if np.average(annotation_attribute) != 0: #safeguard for when LLM gave the same word/phrase different annotations
                print(model1, value, np.average(annotation_attribute))
            else:    
                annotation_material = model_df[model2].loc[value]['materjal'] #take LLM's annotation for this value from elative_material experiment
                if np.average(annotation_material) == 1: #if LLM annotated value as being a material
                    model_results.append('elative_material') #tag the whole row as being an instance of the elative_material construction
                else: #if LLM annotated value as NOT being a material or occupation/nationality/role
                    model_results.append('other') #tag the whole row as being an instance of something other than elative_material or elative_attribute
                    if np.average(annotation_material) != 0: #safeguard for when LLM gave the same word/phrase different annotations
                        print(model2, value, np.average(annotation_attribute))
    return model_results
```

```python
def experiment_small(gold_df, model_df, model, column_gold, column_model, tag_name, gold_list):
    model_results = []
    model_values = model_df[model].index.values.tolist()
    gold_new = []
    i = 0

    for value in gold_df[column_gold]: #iterate over gold standard rows and take a speficied value from each row
        if value in model_values:
            gold_new.append(gold_list[i])
            annotation = model_df[model].loc[value][column_model] #see what annotation the LLM gave the value
            if annotation == 'JAH' or annotation == 1: #if LLM annotated it as the semantic type we want
                model_results.append(tag_name) #give the row that tag
            else: #if LLM annotated it as NOT being the semantic type we want
                model_results.append('other') #tag the whole row as being an instance of something other
        i+=1
    return [gold_new, model_results]
```

```python
def scores(gold_strandard, model_output):
    scorelist = []
    scorelist.append(round(accuracy_score(gold_strandard, model_output),4))
    scorelist.append(round(precision_score(gold_strandard, model_output, average='macro', zero_division=0.0),4))
    scorelist.append(round(recall_score(gold_strandard, model_output, average='macro'),4))
    scorelist.append(round(f1_score(gold_strandard, model_output, average='macro'),4))
    scorelist.append(round(f1_score(gold_strandard, model_output, average='weighted'),4))
    return scorelist
```

### Combine gold-standard labels

```python
annotator1 = list(annotator_1_df['manual_annotation'])
```

```python
#combine gold standard labels into two values: elative_attribute and other
annotator1_attribute = []
annotator1 = list(annotator_1_df['manual_annotation'])
for label in annotator1:
    if label in ['elative_material', 'other', 'error']:
        annotator1_attribute.append('other')
    else:
        annotator1_attribute.append(label)
```

```python
#combine gold standard labels into two values: elative_material and other
annotator1_material = []
for label in annotator1:
    if label in ['elative_attribute', 'other', 'error']:
        annotator1_material.append('other')
    else:
        annotator1_material.append('elative_material')
```

```python
#combine gold standard labels into three values: elative_attribute, elative_material and other
annotator1_all = []
annotator1 = list(annotator_1_df['manual_annotation'])
for label in annotator1:
    if label in ['other', 'error']:
        annotator1_all.append('other')
    elif label == 'elative_attribute':
        annotator1_all.append('elative_attribute')
    elif label == 'elative_material':
        annotator1_all.append('elative_material')
```
